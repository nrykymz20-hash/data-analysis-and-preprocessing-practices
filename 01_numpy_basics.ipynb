```python
# Creating a 1D NumPy array
arr_1D=np.array([10,20,30,40,50,60,70])
```

```python
# Accessing elements in a 1D NumPy array using positive and negative indexing
print(arr_1D[3]) # Returns the element at zero-based index 3
print(arr_1D[-1]) # Returns the last element of the array
print(arr_1D[-3]) # Returns the third element from the end
```

```python
# 1D Array Slicing Syntax: [start : stop : step]
# Note: 'stop' index is always excluded (upper bound is not included)
print(arr_1D[1:5])  # Output: [20 30 40 50] | Slices from index 1 up to (excluding) index 5
print(arr_1D[:5])   # Output: [10 20 30 40 50] | Slices from index 0 up to (excluding) index 5
print(arr_1D[4:])   # Output: [50 60 70] | Slices from index 4 all the way to the end
print(arr_1D[::])   # Output: [10 20 30 40 50 60 70] | Returns a complete copy/view of the entire array
print(arr_1D[1:6:3]) # Output: [20 50] | Slices from index 1 to 6 with a step size of 3
print(arr_1D[::2])  # Output: [10 30 50 70] | Select every 2nd element from start to the end
```

```python
# Creating a 2D NumPy array
arr_2D=np.array([
                    [10,20,30],
                    [40,50,60],
                    [70,80,90]
                ])
print(arr_2D)
```

```python
# 2D Array Slicing Syntax: name_array[row_slice, col_slice]
# Note: Syntax follows [start:stop:step]. Upper bound ('stop') is always excluded.
print(arr_2D[1, 2]) # Accessing row index 1, column index 2
print(arr_2D[0,1])  # Accessing row index 0, column index 1
print(arr_2D[:,1])  # Extracting all rows from column index 1 (the entire 2nd column)
print(arr_2D[:,0:2]) # Extracting all rows for column indices 0 and 1 (first 2 columns)
print(arr_2D[0:2,0:2]) # Extracting a sub-matrix: rows 0 and 1, columns 0 and 1 (top-left 2x2 section)
```

```python
# Accessing elements using Boolean Masking (requires square brackets)
print(arr_2D[arr_2D>15]) # Returns elements greater than 15
print(arr_2D[arr_2D<10]) # Returns elements lesser than 10
print(arr_2D[arr_2D>=50]) # Returns elements greater or equals to 50
```

```python
# Creating a 3D NumPy array
arr_3D=np.array([
        [
            [12,52,33],
            [40,5,16]
        ],
        [
            [18,9,12],
            [1,37,44]
        ],
        [
            [4,29,53],
            [31,90,78]
        ]

])
```

```python
#Accessing elements of 3D NumPy array
print(arr_3D[0]) # Accessing the entire 1st matrix (depth index 0)
print(arr_3D[2]) # Accessing the entire 3rd matrix (depth index 2)
print(arr_3D[1,1]) # Extracting a specific row: 2nd matrix (index 1), 2nd row (index 1)
print(arr_3D[0,1,2]) # Accessing a single element: 1st matrix (index 0), 2nd row (index 1), 3rd column
```

```python
# NumPy Data Types (dtypes)

# 1. Automatic Type Inference:
# NumPy automatically detects integer input and assigns the default integer type (int64/int32 depending on OS)
array1=np.array([7,11,34])
print("Data Type of array1 ",array1.dtype)
```

```python
# 2. Explicit Type Definition:
# Explicitly specifying the data type to optimize memory usage (32-bit integer).
array2=np.array([1,5,32],dtype=np.int32)
print("Data Type of array2 ",array2.dtype)
```

```python
array3=np.array([1,2,8.5,3.2],dtype=np.float32) # Explicitly creating a floating-point array with 32-bit precision (float32)
print("Data Type of array3 ",array3.dtype)
```

```python
# Precision comparison between 32-bit (float32) and 64-bit (float64) floating-point numbers
array4=np.array([7.1234567891011],dtype=np.float32)
array5=np.array([7.1234567891011],dtype=np.float64)
print("array4 (float32): ",array4) # float32 loses precision after ~7 decimal places
print("array5 (float64): ",array5) # float64 preserves precision up to ~15-17 decimal places
```
